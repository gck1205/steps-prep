```python
# input_lst = [(1,2), (3,2), (4,2), (1,2), (3,2), (5,2), (1,2), (5,2)]
input_lst=[[1,2],[3,2],[4,2],[1,2],[3,2],[5,2],[1,2],[5,2]]
# Using sets to find all unique duplicates
output_lst=[tuple(i) for i in input_lst]
seen = set()
duplicates = set()

for item in output_lst:
    if item in seen:
        duplicates.add(item)
    else:
        seen.add(item)

# Convert the set of duplicates back to a list (optional, for specific output format)
list_of_duplicates = [list(item) for item in duplicates]

print(f"The duplicates are: {list_of_duplicates}")
```
