```python
from kafka import KafkaProducer,KafkaConsumer
import time
import random
import datetime
import json
import uuid
# from generate_data import gen_event
```

```python
event_status=["SUCCESS","ERROR","STANDBY"]+[None]
device_id=["d_"+str(id).rjust(4,'0') for id in range(1,6)]
customer_id=["CI"+str(id).rjust(5,'0') for id in range(100,121)]
```

```python
def gen_event(offset):
    event={
    "eventId":str(uuid.uuid4()),
    "eventOffset":offset,
    "eventPublisher":"device",
    "customerId":random.choice(customer_id),
    "data": {
        "devices": [
            {
                "deviceId":random.choice(device_id),
                "temperature":random.randint(30,70),
                "measure":"C",
                "status":random.choice(event_status)
            } for i in range(random.randint(1,3))
        ],
    },
    "eventTime":str(datetime.datetime.now())
    }
    return json.dumps(event)
```

```python
bs_server="ed-kafka:29092"
topic="device-data"
```

```python
def post_to_kafka(data):
    print("data"+str(data))
    producer=KafkaProducer(bootstrap_servers=bs_server)
    producer.send(topic,key=bytes(str(uuid.uuid4()),'utf-8'),value=data)
    producer.close()
    print("posted to topic")
```

```python
if __name__ == "__main__":
   offset=25
   while (offset>0):
       print(gen_event(offset))
       post_to_kafka(bytes(str(gen_event(offset)),'utf-8'))
       offset=offset-1
       time.sleep(5)
```
